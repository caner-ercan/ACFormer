```python
import os

def check_corresponding_files(parent_folder1, parent_folder2):
    # Iterate through each subfolder in the first parent folder
    for subfolder in os.listdir(parent_folder1):
        subfolder_path1 = os.path.join(parent_folder1, subfolder)
        subfolder_path2 = os.path.join(parent_folder2, subfolder)

        # Check if the subfolder exists in the second parent folder
        if os.path.isdir(subfolder_path1) and os.path.isdir(subfolder_path2):
            # Get the list of files in both subfolders
            files1 = os.listdir(subfolder_path1)
            files2 = os.listdir(subfolder_path2)

            # Create sets for easy comparison
            # files1_set = {os.path.splitext(file)[0]: file for file in files1}
            files1_set = [file for file in files1]
            files2_set = [os.path.splitext(file)[0] for file in files2]

            # Check for corresponding files
            for file1 in files1_set:
                if not file1 in files2_set:
                    print(f"No match for: {file1} in {subfolder_path2}")

```

```python

# Example usage
parent_folder1 = "/rsrch5/home/trans_mol_path/cercan/data/BE/cell_detect/acformer/NU_dataset_rois//tiles/"
parent_folder2 = "/rsrch5/home/trans_mol_path/cercan/data/BE/cell_detect/acformer/NU_dataset_rois//predictions/"
check_corresponding_files(parent_folder1, parent_folder2)
```
