```python
import os
import random
random.seed(0)
import shutil
```

```python
parent_folder = "/Volumes/rsrch5/home/trans_mol_path/cercan/data/BE/cell_detect/acformer/finetune/qpproj_batch2/finetune_annotations"

tile_source_folder = os.path.join(parent_folder,"tiles")
csv_source_folder = os.path.join(parent_folder,"corrected_point")
slide_list = os.listdir(tile_source_folder)
random.shuffle(slide_list)
core_slide_list = [value[:5] for value in slide_list]

training_wsi = core_slide_list[:int(len(core_slide_list)*0.8)]
val_wsi = core_slide_list[int(len(core_slide_list)*0.8):]

training_wsi_full = slide_list[:int(len(core_slide_list)*0.8)]
val_wsi_full = slide_list[int(len(core_slide_list)*0.8):]
```

```python
def copy_cvs(core_slide_list,csv_source_folder,csv_destination_folder):
    for filename in os.listdir(csv_source_folder):
        if any(filename.startswith(core_wsi_name) for core_wsi_name in core_slide_list):
            source_csv = os.path.join(csv_source_folder, filename )
            shutil.copy2(source_csv, csv_destination_folder)

csv_destination_folder = os.path.join(parent_folder,"training","csv")
copy_cvs(training_wsi,csv_source_folder,csv_destination_folder)

csv_destination_folder = os.path.join(parent_folder,"validation","csv")
copy_cvs(val_wsi,csv_source_folder,csv_destination_folder)
```

```python
def copy_tile(full_slide_list,tile_source_folder,tile_destination_folder):
    for slide_full_name in full_slide_list:
        slide_tile_folder = os.path.join(tile_source_folder,slide_full_name)
        for file in os.listdir(slide_tile_folder):
            shutil.copy2(os.path.join(slide_tile_folder,file), tile_destination_folder)

tile_destination_folder = os.path.join(parent_folder,"training","tile")
copy_tile(training_wsi_full,tile_source_folder,tile_destination_folder)

tile_destination_folder = os.path.join(parent_folder,"validation","tile")
copy_tile(val_wsi_full,tile_source_folder,tile_destination_folder)
```
